# Merging of dataframes

```python
# merge(): this function can be used to merge the dataframes based upon common column
# merge(): function work like join in sql : inner, left, right outer

df1 = pd.DataFrame({"ID" : [1,2,3],"Dream_company" : ["google","microsoft","Apple"]})
df2 = pd.DataFrame({"ID" : [2,3,4], "Dream_Package" : [10,15,18]})


#inner Join
merge_df = pd.merge(df1,df2, on = "ID")
merge_df


#outer Join
```

```python
#left join
merge_df = pd.merge(df1,df2, on = "ID", how = "left")
merge_df
```

```python
merge_df = pd.merge(df1,df2,on= "ID", how = "right")
merge_df
```

```python
merge_df = pd.merge(df1,df2,on= "ID", how = "outer")
merge_df
```

# Join() : Mainly used to work with index

```python
df3 = pd.DataFrame({"Hero_Name": ["NTR","Prabhas","Pawan"]}, index = [1,2,3])
df4 = pd.DataFrame({"Heroine_Name": ["Kajol","Anushkha","Shurti"]}, index = [2,3,4])
joined_df = df3.join(df4, how = "inner")
joined_df
```

```python
joined_df = df3.join(df4, how="outer")
joined_df
```

# concat

```python
# Def : concat() function is used to stack dataframes vertically or horizontally 
```

```python
df5 = pd.DataFrame({"ID": [1,2],"Fav_Movie" : ["Paradise","Bahuballi"]})
df6 = pd.DataFrame({"ID" : [3,4], "Fav_Movie" : ["Spiderman","Doomsday"]})
df7 = pd.DataFrame({"ID": [5,6], "Fav_Movie": ["Pushpa","Maghadeera"]})


# axis = 0 : Veritically (adding rows)
# axis = 1 : Horizontally (adding columns)
concat_df = pd.concat([df5,df6], axis = 0, ignore_index = True)
concat_df
```

```python
concat_df = pd.concat([df5,df6], axis = 1, ignore_index = True)
concat_df
```

```python
concat_df = pd.concat([df5,df6, df7], axis = 0, ignore_index = True)
concat_df
```

```python
concat_df = pd.concat([df5,df6,df7], axis = 1, ignore_index = True)
concat_df
```
